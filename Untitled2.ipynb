```python
!pip install -q kagglehub
import kagglehub, os, pandas as pd, numpy as np
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")
orders = pd.read_csv(f"{path}/olist_orders_dataset.csv")
payments = pd.read_csv(f"{path}/olist_order_payments_dataset.csv")
orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
print("Loaded:" , orders.shape ,payments.shape)
```

```python
df = orders.merge(payments, on='order_id')
df = df[df['order_status']=='delivered']
snapshot = df['order_purchase_timestamp'].max() + pd.Timedelta(days=1)

rfm = df.groupby('customer_id').agg(
    Recency=('order_purchase_timestamp',lambda x: (snapshot - x.max()).days),
    Frequency=('order_id', 'nunique'),
    Monetary=('payment_value', 'sum')
).reset_index()

print(rfm.head())
print("RFM shape:",rfm.shape)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,3))
plt.subplot(1,3,1); sns.histplot(rfm['Recency'], bins=30, color='blue'); plt.title('Recency - Days since last order')
plt.subplot(1,3,2); sns.histplot(rfm['Frequency'],bins=10, color='green'); plt.title('Frequency - Orders count')
plt.subplot(1,3,3); sns.histplot(rfm['Monetary'],bins=30, color='orange');plt.title('Monetary - Total spent')
plt.tight_layout(); plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np

# Log transform + Scaling - important step
scaled = StandardScaler().fit_transform(np.log1p(rfm[['Recency', 'Frequency', 'Monetary']]))

for k in [3,4,5]:
  km = KMeans(n_clusters=k, random_state=42, n_init=10).fit(scaled)
  score = silhouette_score(scaled, km.labels_)
  print(k, score)
```

```python
final_kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
rfm['Cluster'] = final_kmeans.fit_predict(scaled)

profile = rfm.groupby('Cluster')[['Recency','Frequency','Monetary']].mean().round(1)
print(profile)
print("\ncount:\n", rfm['Cluster'].value_counts())
```

```python
import joblib
joblib.dump(final_kmeans, 'kmeans_model.pk1')
rfm.to_csv('my_segments.csv', index=False)

# Final plot for report
import matplotlib.pyplot as plt
plt.scatter(rfm['Recency'], rfm['Monetary'], c=rfm['Cluster'], cmap='viridis', alpha=0.5)
plt.xlabel('Recency'); plt.ylabel('Monetary'); plt.title('Customer Segments (k=3, Silhouetee=0.388)')
plt.show()
```
